```python
import os
import shutil

src = "/content/drive/MyDrive/GEDI_native_resolution_patches_32x32_11_6"
dst = "/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/data_32_32_patches_11_4_25"

# Create destination directory if it doesn't exist
os.makedirs(dst, exist_ok=True)

# Loop through files in the source directory
for file_name in os.listdir(src):
    if file_name.lower().endswith(".tif"):
        src_path = os.path.join(src, file_name)
        dst_path = os.path.join(dst, file_name)

        if not os.path.exists(dst_path):
            shutil.move(src_path, dst_path)
            print(f"Moved: {file_name}")
        else:
            print(f"Skipped (duplicate): {file_name}")

print("Done!")
```

```python
import os
from collections import Counter

folder = "/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/data_32_32_patches_11_4_25"

# List only files (not subfolders)
files = [f for f in os.listdir(folder) if os.path.isfile(os.path.join(folder, f))]

# Total file count
total_files = len(files)

# Extract prefixes (before first underscore)
prefixes = [f.split("_")[0] for f in files]

# Count occurrences of each prefix
prefix_counts = Counter(prefixes)

print("Total files:", total_files)
print("\nCounts by prefix:")
for prefix, count in sorted(prefix_counts.items()):
    print(f"{prefix}: {count}")
```

```python
import os
from collections import defaultdict

folder = "/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/data_32_32_patches_11_4_25"

# Dictionary: filename -> list of full paths
name_map = defaultdict(list)

for root, dirs, files in os.walk(folder):
    for f in files:
        name_map[f].append(os.path.join(root, f))

# Find duplicates
duplicates = {name: paths for name, paths in name_map.items() if len(paths) > 1}

# Print results
if duplicates:
    print("Duplicate file names found:\n")
    for name, paths in duplicates.items():
        print(f"File name: {name}")
        for p in paths:
            print(f"  - {p}")
        print()
else:
    print("No duplicate filenames found.")
```

```python
import os

folder = "/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/data_32_32_patches_11_4_25"

for fname in os.listdir(folder):
    if "landsat_idx" in fname:
        old_path = os.path.join(folder, fname)
        new_fname = fname.replace("landsat_idx", "landsatidx")
        new_path = os.path.join(folder, new_fname)

        # If the target name exists, add a numeric suffix to avoid collision
        if os.path.exists(new_path):
            base, ext = os.path.splitext(new_fname)
            counter = 1
            while True:
                new_fname2 = f"{base}_{counter}{ext}"
                new_path2 = os.path.join(folder, new_fname2)
                if not os.path.exists(new_path2):
                    new_fname = new_fname2
                    new_path = new_path2
                    break
                counter += 1

        os.rename(old_path, new_path)
        print(f"Renamed: {fname} → {new_fname}")

print("Done!")
```
